```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(0,5,0.1)
print(x)
```

```python
y = np.sin(x)
print(y)
```

```python
plt.plot(x,y)
plt.show()
```

```python
x = np.arange(0,5,0.1)
print(x)
```

```python
Xn=[1,2,3,4]
n = np.arange(0, len(Xn))
print(n)
```

```python
plt.stem(n,Xn)
plt.show()
```

```python
Hn = [1, 1, 0]
n = np.arange(0, len(Hn))
plt.stem(n, Hn)
plt.show()
```

```python
Yn = np.convolve(Xn, Hn)
print(Yn)
```

```python
n = np.arange(0, len(Yn))
plt.stem(n, Yn)
plt.show()
```
